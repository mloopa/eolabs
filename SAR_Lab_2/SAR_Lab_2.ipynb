## 2. Helper functions

These functions keep the notebook readable and reduce repeated plotting code.


```python
def read_raster(path, band=1, downsample=DOWNSAMPLE):
    """Read a single-band raster and return array + metadata.
    Nodata values are converted to NaN.
    """
    with rasterio.open(path) as src:
        nodata = src.nodata
        if downsample > 1:
            new_h = max(1, src.height // downsample)
            new_w = max(1, src.width // downsample)
            arr = src.read(
                band,
                out_shape=(new_h, new_w),
                resampling=Resampling.average,
            ).astype("float32")
            transform = src.transform * Affine.scale(
                src.width / new_w, src.height / new_h
            )
        else:
            arr = src.read(band).astype("float32")
            transform = src.transform
        meta = src.meta.copy()
        meta.update({"height": arr.shape[0], "width": arr.shape[1]})
        bounds = src.bounds
        crs = src.crs

    if nodata is not None:
        arr[arr == nodata] = np.nan
    return arr, meta, bounds, transform, crs
    

def _mem_mb(arr):
    return arr.nbytes / 1024**2


def plot_image(arr, title="", vmin=None, vmax=None, cmap="gray", figsize=(8, 6)):
    fig, ax = plt.subplots(figsize=figsize)
    im = ax.imshow(arr, cmap=cmap, vmin=vmin, vmax=vmax)
    fig.colorbar(im, ax=ax)
    ax.set_title(title)
    ax.axis("off")
    plt.show()
    plt.close(fig)  # release figure memory immediately


def plot_side_by_side(arr1, arr2, title1, title2, vmin=None, vmax=None, cmap="gray", figsize=(14, 6)):
    fig, axes = plt.subplots(1, 2, figsize=figsize)
    axes[0].imshow(arr1, cmap=cmap, vmin=vmin, vmax=vmax)
    axes[0].set_title(title1)
    axes[0].axis("off")
    im2 = axes[1].imshow(arr2, cmap=cmap, vmin=vmin, vmax=vmax)
    axes[1].set_title(title2)
    axes[1].axis("off")
    fig.colorbar(im2, ax=axes.ravel().tolist(), shrink=0.8)
    plt.show()
    plt.close(fig)  # release figure memory immediately


def to_db(arr, eps=1e-6):
    """Convert linear power to dB. Skip if rasters are already in dB."""
    return 10 * np.log10(np.maximum(arr, eps))


def robust_limits(arr, pmin=2, pmax=98):
    """Percentile-based display limits, ignoring NaN."""
    return np.nanpercentile(arr, pmin), np.nanpercentile(arr, pmax)


def histogram(arr, bins=100, title="", xlabel="", logy=False, max_samples=2_000_000):
    """Plot a histogram without duplicating the entire array in memory.

    For very large arrays we draw a random sample (max_samples pixels)
    so np.histogram never has to hold a full copy of the data.
    """
    flat = arr.ravel()
    finite_idx = np.where(np.isfinite(flat))[0]
    if len(finite_idx) > max_samples:
        rng = np.random.default_rng(42)
        finite_idx = rng.choice(finite_idx, size=max_samples, replace=False)
    valid = flat[finite_idx]
    del flat, finite_idx  # free immediately

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.hist(valid, bins=bins)
    del valid
    if logy:
        ax.set_yscale("log")
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Pixel count")
    ax.grid(True, alpha=0.3)
    ax.set_xlim(xmin=0, xmax=200)
    plt.show()
    plt.close(fig)


def pick_points(image, title="Click 2 points, then close the figure", cmap="gray", vmin=None, vmax=None):
    """Display an image and let the user click two points."""
    fig, ax = plt.subplots(figsize=(10, 8))
    ax.imshow(image, cmap=cmap, vmin=vmin, vmax=vmax)
    ax.set_title(title)
    ax.set_xlabel("X pixel")
    ax.set_ylabel("Y pixel")
    pts = plt.ginput(2, timeout=-1)
    plt.close(fig)
    return [(int(round(x)), int(round(y))) for x, y in pts]


def extract_transect(arr, p0, p1, n=500):
    """Extract pixel values along the line between two coordinates."""
    x0, y0 = p0
    x1, y1 = p1
    xs_i = np.clip(np.round(np.linspace(x0, x1, n)).astype(int), 0, arr.shape[1] - 1)
    ys_i = np.clip(np.round(np.linspace(y0, y1, n)).astype(int), 0, arr.shape[0] - 1)
    return np.arange(n), arr[ys_i, xs_i], xs_i, ys_i


def plot_transect_on_image(image, p0, p1, title="", cmap="gray", vmin=None, vmax=None):
    """Overlay a transect line on an image."""
    fig, ax = plt.subplots(figsize=(10, 8))
    ax.imshow(image, cmap=cmap, vmin=vmin, vmax=vmax)
    ax.plot([p0[0], p1[0]], [p0[1], p1[1]], 'r-', linewidth=2)
    ax.plot(p0[0], p0[1], 'go', markersize=8, label='Start')
    ax.plot(p1[0], p1[1], 'bo', markersize=8, label='End')
    ax.set_title(title)
    ax.set_xlabel("X pixel")
    ax.set_ylabel("Y pixel")
    ax.legend()
    plt.show()
    plt.close(fig)
```

**QUESTION (10pts): Why is the CRS (coordinate reference system) for the Pre-TC view "None" while the post-TC has "EPSG:4326"?**

**ANSWER:** The pre-TC image is stored in slant-range / range-Doppler coordinates — pixels are indexed by radar range sample (across-track time delay) and azimuth line (along-track pulse index). This geometry has no direct mapping to geographic latitude/longitude, so no CRS is assigned. Terrain correction (Range-Doppler geocoding) uses the DEM to project each radar pixel onto its true geographic location on the Earth's surface and resamples the result onto a regular geographic grid. The output is therefore referenced to EPSG:4326 (WGS84 geographic coordinates).

**QUESTION (10pts): Why can't we overlay the Pre-TC image with the DEM? Give two reasons.**

**ANSWER:**

1. pre-TC image has CRS=None, which means there is no geographic coordinate attached to any pixel, so it cannot be matched to DEM
2. geomtric distortions such as foreshortening, layover and shadow would make an overlay with DEM meaningless because they do not share the same spatial geometry

## 10. Incidence angle

If you exported an incidence-angle raster, this is a good place to inspect it and compare it with brightness.

**QUESTION (10pts): Where are bright/dark patterns linked to local geometry rather, and where are bright/dark patterns linked to land cover.**

**ANSWER:**

- **Geometry-driven bright:** steep east-facing slopes with very low local incidence angles appear anomalously bright due to foreshortening/layover — this is a geometry artifact independent of what is actually on the slope.
- **Geometry-driven dark:** shadow zones on west-facing slopes are uniformly near-zero regardless of land cover, because the radar simply never illuminated them.
- **Land-cover-driven:** on valley floors and gentle slopes where the local incidence angle is moderate (~35–45°), brightness differences between forest, bare rock, fields, and water bodies reflect true surface roughness and dielectric properties rather than viewing geometry.

## Reflection question

**QUESTION (10pts): If a researcher downloaded a ready-made GRD product and never thought about terrain effects, what kinds of interpretation errors might they make in a mountainous study area?**

**ANSWER:**
- They would misclassify bright foreshortening/layover zones as high-backscatter land cover (e.g. dense forest or urban areas) when the brightness is purely geometric.
- Shadow zones would be misidentified as water bodies or smooth surfaces, hiding whatever land cover is actually there.
- Features would be geographically mislocated due to slant-range displacement, causing false misregistration with maps or other datasets.
- Spurious land-cover change would be detected when comparing images from slightly different look angles, since the geometry-driven brightness differences would appear as real change.
- Per-pixel area or biomass estimates would be biased because the ground footprint of each pixel varies with local slope angle.
